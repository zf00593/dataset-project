```python
import pandas as pd

df = pd.read_excel('extracted_files/Cyber_Awareness_Survey.xlsx')
```

```python
# Count totals of each column and map yes to 1 and no to 0
p_change_agg = df["Do you change your password Frequently?"].map({1:"Yes", 0:"No"}).value_counts()

# Count totals of cyber awareness and map yes to 1 and no to 0
cyber_aware_agg = df["Are you aware of cyber laws?!"].map({1:"Yes", 0:"No"}).value_counts()

# Count totals of how frequent the password is changed
p_change_freq = df["How Frequent!?"].value_counts()

# Count totals of social media usage for each distinct amount of time
social_media_counts = df["Social Media Usage (in Hours)"].value_counts()

# Count the totals for each social media platform
top_social_media = df["Preferred Social Media 1"].value_counts()

# Create a cross table of how frequent password is changed and if the individual thinks they change the password frequently
corr_freq_chg = pd.crosstab(
    df["Do you change your password Frequently?"].map({1:"Yes", 0:"No"}),
    df["How Frequent!?"]
)

# Cross table between social media usage and their preferred platform
usage_plat = pd.crosstab(
    df["Preferred Social Media 1"],
    df["Social Media Usage (in Hours)"]
)

# Cross table between social media usage and if the individual posts personal pictures or shares important events
usage_personal = pd.crosstab(
    df["Do you post personal pictures, share important events of your life in Social Media including Messenger Apps?"].map({2:"Post personal pictures and share important events",1:"Post just personal pictures", 0:"None"}),
    df["Social Media Usage (in Hours)"]
)

# Cross table between social media usage and if the individual posts personal pictures or shares important events, normalised by columns into proportions
usage_personal_norm = pd.crosstab(
    df["Do you post personal pictures, share important events of your life in Social Media including Messenger Apps?"]
      .map({2:"Pictures + events", 1:"Pictures only", 0:"None"}),
    df["Social Media Usage (in Hours)"],
    normalize="columns"
) * 100
```

```python
def weighted_ranking(df):
    """Calculates a weighted ranking of each platform based on the preference each person put the application as

    Args:
        df (pd.DataFrame): A dataframe with the columns "Preferred Social Media 1", "Preferred Social Media 2", "Preferred Social Media 3", and "Preferred Social Media 4"

    Returns:
        pd.DataFrame: A dataframe with the platforms and their weighted score sorted in descending order
    """
    weights = {
        "Preferred Social Media 1": 4,
        "Preferred Social Media 2": 3,
        "Preferred Social Media 3": 2,
        "Preferred Social Media 4": 1
    }

    scores = {}

    # Loops over each column and its weight
    for col, weight in weights.items():
        # Loops through all platforms in that column
        for row_platform in df[col]:
            # Makes a dictionary of platforms and gets the score for the platform defaulting to 0 and then adds the weight
            scores[row_platform] = scores.get(row_platform, 0) + weight

    weighted_ranking = (
        pd.DataFrame(scores.items(), columns=["Social Media", "Weighted Score"])
        .sort_values("Weighted Score", ascending=False)
    )

    return weighted_ranking
```
