```python
def weighted_ranking(df):
    weights = {
        "Preferred Social Media 1": 4,
        "Preferred Social Media 2": 3,
        "Preferred Social Media 3": 2,
        "Preferred Social Media 4": 1
    }

    scores = {}

    # Loops over each column and its weight
    for col, weight in weights.items():
        # Loops through all platforms in that column
        for platform in df[col]:
            
            scores[platform] = scores.get(platform, 0) + weight

    weighted_ranking = (
        pd.DataFrame(scores.items(), columns=["Social Media", "Weighted Score"])
        .sort_values("Weighted Score", ascending=False)
    )

    return weighted_ranking
```
