### Visualisation code:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

analysis = SocialMediaAnalysis()
```

Password change habits:

```python
password_changes = analysis.password_change_summary()

plt.figure(figsize=(4, 4))

password_changes.plot(kind="bar", color="#33D5FA", legend=False, width=0.3)

plt.title("Do Respondents Frequently Change Their Password?",fontsize=10, fontweight="bold")
plt.xlabel("Response", fontsize=6)
plt.ylabel("Number of Respondents", fontsize=6)
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
password_cyber = pd.crosstab(
    analysis.df["How Frequent!?"],
    analysis.df["Are you aware of cyber laws?!"].map({
        1: "Aware",
        0: "Not Aware"
    })
)

password_cyber.plot(
    kind="bar",
    figsize=(10, 6),
    color=["#339CF5", "#056FCC"]
)

plt.title("Password Change Frequency vs Cyber-Law Awareness")
plt.xlabel("Password Change Frequency")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=45)
plt.legend(title="Cyber-Law Awareness")

plt.tight_layout()
plt.show()
```

```python
ranking = analysis.weighted_ranking()

plt.figure(figsize=(9, 5))

plt.barh(
    ranking["Social Media"],        
    ranking["Weighted Score"],      
    color="#339CF5"
)

plt.title("Overall Social Media Ranking")
plt.xlabel("Weighted Score")
plt.ylabel("Social Media Platform")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 8))

wedges, texts, autotexts = plt.pie(
    ranking["Weighted Score"],
    labels=ranking["Social Media"],
    autopct="%1.1f%%",
    startangle=90,
    labeldistance=1.15,
    pctdistance=0.75,
    wedgeprops=dict(width=0.4),
    colors = ["#04768F", "#B9F1FD", "#023946", "#33D5FA"],
    textprops={"fontsize": 11}
    
)

for text in texts:
    text.set_fontsize(12)

for autotext in autotexts:
    autotext.set_fontsize(10)
    autotext.set_color("white")
    
plt.title(
    "Weighted Distribution of Social Media Preferences",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
privacy_cyber = pd.crosstab(
    analysis.df["Are you aware of cyber laws?!"].map({
        1: "Aware",
        0: "Not Aware"
    }),
    analysis.df["Do you enable/modify privacy settings before you post any message?"].map({
        1: "Yes",
        0: "No"
    }),
    normalize="index"
) * 100

ax = privacy_cyber.plot(
    kind="bar",
    stacked=True,
    figsize=(7, 5),
    color=["#04768F", "#60DEFB"]
)

plt.title("Cyber-Law Awareness vs Use of Privacy Settings")
plt.xlabel("Cyber-Law Awareness")
plt.ylabel("Percentage of Respondents")
plt.xticks(rotation=0)

plt.legend(
    title="Uses Privacy Settings",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```
