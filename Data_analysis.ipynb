```python
import zipfile 

with zipfile.ZipFile('archive.zip', 'r') as zip_ref:
    zip_ref.extractall('extracted_files')
    
```

```python
class SocialMediaAnalysis:

    def __init__(self):
        import pandas as pd
        self.df = pd.read_excel('extracted_files/Cyber_Awareness_Survey.xlsx')

    def password_change_summary(self):
        """
        Returns a summary of password change habits.

        Returns:
            pd.Series: A series with the counts of each password change response.
        """
        return (
            self.df["Do you change your password Frequently?"]
            .map({1: "Yes", 0: "No"})
            .value_counts()
        )

    def cyber_awareness_summary(self):
        """
        Returns a summary of cyber awareness.

        Returns:
            pd.Series: A series with the counts of each cyber awareness response.
        """
        return (
            self.df["Are you aware of cyber laws?!"]
            .map({1: "Yes", 0: "No"})
            .value_counts()
        )

    def password_change_frequency(self):
        """
        Returns a summary of how frequently the password is changed.

        Returns:
            pd.Series: A series with the counts of each password change frequency.
        """
        return self.df["How Frequent!?"].value_counts()

    def social_media_usage_summary(self):
        """
        Returns a summary of social media usage for each distinct amount of time.

        Returns:
            pd.Series: A series with the counts of each social media usage duration.
        """
        return self.df["Social Media Usage (in Hours)"].value_counts()

    def preferred_social_media_summary(self):
        """
        Returns a summary of the preferred social media platform.

        Returns:
            pd.Series: A series with the counts of each preferred social media platform.
        """
        return self.df["Preferred Social Media 1"].value_counts()

    def password_change_frequency_correlation(self):
        """
        Returns a cross table of how frequent the password is changed and if the individual thinks they change the password frequently.

        Returns:
            pd.DataFrame: A dataframe with the cross tabulated values.
        """
        return pd.crosstab(
            self.df["Do you change your password Frequently?"]
            .map({1: "Yes", 0: "No"}),
            self.df["How Frequent!?"]
        )

    def platform_vs_usage(self):
        """
        Returns a cross table between social media usage and their preferred platform.

        Returns:
            pd.DataFrame: A dataframe with the cross tabulated values.
        """
        return pd.crosstab(
            self.df["Preferred Social Media 1"],
            self.df["Social Media Usage (in Hours)"]
        )

    def personal_posting_vs_usage(self):
        """
        Returns a cross table between social media usage and if the individual posts personal pictures or shares important events.

        Returns:
            pd.DataFrame: A dataframe with the cross tabulated values.
        """
        return pd.crosstab(
            self.df[
                "Do you post personal pictures, share important events of your life in Social Media including Messenger Apps?"
            ].map({
                2: "Post personal pictures and share important events",
                1: "Post just personal pictures",
                0: "None"
            }),
            self.df["Social Media Usage (in Hours)"]
        )

    def personal_posting_vs_usage_normalised(self):
        """
        Returns a cross table between social media usage and if the individual posts personal pictures or shares important events, normalised by columns into proportions.

        Returns:
            pd.DataFrame: A dataframe that looks like this
        """
        social_individual_normalised =  pd.crosstab(
                        self.df[
                            "Do you post personal pictures, share important events of your life in Social Media including Messenger Apps?"
                        ].map({
                            2: "Pictures + events",
                            1: "Pictures only",
                            0: "None"
                        }),
                        self.df["Social Media Usage (in Hours)"],
                        normalize="columns"
        ) * 100
        
        return social_individual_normalised
    
    
    def weighted_ranking(self):
        """Calculates a weighted ranking of each platform based on the preference each person put the application as

        Args:
            df (pd.DataFrame): A dataframe with the columns "Preferred Social Media 1", "Preferred Social Media 2", "Preferred Social Media 3", and "Preferred Social Media 4"

        Returns:
            pd.DataFrame: A dataframe with the platforms and their weighted score sorted in descending order
        """
        
        weights = {
            "Preferred Social Media 1": 4,
            "Preferred Social Media 2": 3,
            "Preferred Social Media 3": 2,
            "Preferred Social Media 4": 1
        }

        scores = {}

        # Loops over each column and its weight
        for col, weight in weights.items():
            # Loops through all platforms in that column
            for row_platform in self.df[col]:
                # Makes a dictionary of platforms and gets the score for the platform defaulting to 0 and then adds the weight
                scores[row_platform] = scores.get(row_platform, 0) + weight

        weighted_ranking = (
            pd.DataFrame(scores.items(), columns=["Social Media", "Weighted Score"])
            .sort_values("Weighted Score", ascending=False)
        )
        return weighted_ranking
        
```
